```python
import pandas as pd

# modeled data path.
modeled_path = "../datasets/03_gold_modeled_data/"
```

```python
fact_sales = pd.read_parquet(f"{modeled_path}fact_sales.parquet")
dim_orders = pd.read_parquet(f"{modeled_path}dim_orders.parquet")
dim_sellers = pd.read_parquet(f"{modeled_path}dim_sellers.parquet") 
dim_products = pd.read_parquet(f"{modeled_path}dim_products.parquet")
dim_customers = pd.read_parquet(f"{modeled_path}dim_customers.parquet")
dim_location = pd.read_parquet(f"{modeled_path}dim_location.parquet")
```

```python
# Validate the relationships, data integrity, referential and PK constraints.
assert (fact_sales["order_key"].isin(dim_orders["order_key"]).all()) & (dim_orders["order_id"].is_unique)
assert (fact_sales["product_key"].isin(dim_products["product_key"]).all()) & (dim_products["product_id"].is_unique)
assert (fact_sales["seller_key"].isin(dim_sellers["seller_key"]).all()) & (dim_sellers["seller_id"].is_unique)
assert (dim_orders["customer_key"].isin(dim_customers["customer_key"]).all() & dim_customers["customer_key"].is_unique)
assert (dim_customers["location_key"].isin(dim_location["location_key"]).all() & dim_location["location_key"].is_unique)
assert (dim_sellers["location_key"].isin(dim_location["location_key"])).all()
```
